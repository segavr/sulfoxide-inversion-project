```python
from rdkit import Chem
from rdkit.Chem import AllChem

# Create methyl phenyl sulfoxide from SMILES
mol = Chem.MolFromSmiles('CS(=O)c1ccccc1')
mol = Chem.AddHs(mol)

# Generate 3D coordinates and optimize with MMFF
AllChem.EmbedMolecule(mol, randomSeed=42)
AllChem.MMFFOptimizeMolecule(mol)

conf = mol.GetConformer()
atoms = mol.GetAtoms()

# Save XYZ geometry to ../data/methylphenylsulfoxide.xyz
with open('../data/methylphenylsulfoxide.xyz', 'w') as f:
    f.write(f'{mol.GetNumAtoms()}\n')
    f.write('methylphenylsulfoxide initial geometry (RDKit MMFF)\n')
    for atom in atoms:
        pos = conf.GetAtomPosition(atom.GetIdx())
        f.write(f'{atom.GetSymbol():2s}  {pos.x:12.6f}  {pos.y:12.6f}  {pos.z:12.6f}\n')

print("Done, XYZ file saved.")
print("Number of atoms:", mol.GetNumAtoms())
```
